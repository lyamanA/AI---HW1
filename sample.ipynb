```python
"""
Задание 1
Создайте двумерный массив размером 4×4, заполненный случайными целыми
числами от 1 до 50.
• Выведите массив в читаемом виде (в виде таблицы).
• Найдите сумму элементов по каждой строке и по каждому столбцу
(используйте параметр axis).
• Определите общее среднее значение массива.
• Выведите все результаты в удобном виде (например: «Суммы по строкам: …»,
«Суммы по столбцам: …»).
"""

import numpy as np

array = np.random.randint(1, 51, size=(4, 4))

print(array)
```

```python
# Суммы по строкам
row_sums = np.sum(array, axis=1)
print("Суммы по строкам:",row_sums)
```

```python
# Суммы по столбцам
col_sums = np.sum(array, axis=0)
print("Суммы по столбцам:",col_sums)
```

```python
#общее среднее значение массива
average = np.mean(array)
print("Общее среднее значение массива:",average)
```

```python
"""
Задание 2
Создайте массив размером 3×5, заполненный числами от 0 до 14.
• Выведите исходный массив.
• Выполните транспонирование массива (строки становятся столбцами, а
столбцы строками).
• Примените операцию reshape, чтобы получить массив 5×3.
• Отобразите результаты на экран.
"""

array_2 = np.arange(15).reshape((3, 5))
print("\nИсходный массив:\n",array_2)
```

```python
# Транспонирование массива
transposed_array = array_2.T
print("\nТранспонированный массив:\n",transposed_array)
```

```python
reshaped_array = array_2.reshape(5, 3)
print("\nМассив после reshape:\n",reshaped_array)
```

```python
"""
Задание 3
Создайте Маrkdown-ячейку с заголовком "Отчёт по продажам" и подпишите: «В этом разделе мы считаем итоги по простым данным».
В кодовой ячейке создайте список продаж за неделю:
sales [120, 135, 100, 150, 90, 200, 170]
Посчитайте:
общее количество продаж за неделю;
среднее значение;
максимальное и минимальное значения.
Создайте Markdown-ячейку и оформите таблицу с результатами
"""

sales = [120, 135, 100, 150, 90, 200, 170]
total_sales = sum(sales)
average_sales = total_sales / len(sales)
max_sales = max(sales)
min_sales = min(sales)

print("Общая сумма:", total_sales)
print("Среднее значение:", average_sales)
print("Максимум:", max_sales)
print("Минимум:", min_sales)
```
